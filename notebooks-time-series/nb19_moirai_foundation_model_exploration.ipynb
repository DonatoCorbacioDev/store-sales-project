# Engineering Notes

This exploratory notebook highlighted an important ML engineering challenge:

modern forecasting foundation models introduce heavy dependency requirements that may conflict with existing forecasting environments.

The installation of uni2ts / Moirai introduced:

- Torch version downgrades
- NumPy downgrades
- GluonTS incompatibilities
- AutoGluon dependency conflicts
- large CUDA/JAX runtime installations

This demonstrates that foundation-model experimentation should generally be isolated in dedicated environments or containers rather than integrated directly into stable production-oriented forecasting pipelines.

The experiment was therefore considered valuable from an engineering perspective even without completing a full forecasting benchmark.
